```python
census = {
"household 1": {
    "name": "Blake",
    "age": 20,
    "income": 100000,
},
"household 2": {
    "name": "Gurt",
    "age": 25,
    "income": 200000,
}

print(census["household 2"]["income"][2])

print(len(census))
print(len(census["household 1"]))
print(len(census["household 1"]["income"]))
```

#### lists allow us to store many values in one olace and we access them by their index number in the list. Dicitonaries allow us to store things by key and we access the value by referencing the key. It is important to keep track of what data tupe we are using at any given time. For example, in the example above, census is a dictionary. census["household2"] is also a dictionary, census["household2"][income] is a list and census ["household2]["income"][2] is an integer
