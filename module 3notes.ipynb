```python
clothing_stores = ["walmart", "pandabuy", "nike", "dior", "adidas"]
fav_store = input("enter your favorite clothing store")
for store in clothing_stores:
    if store.lower() == fav_store.lower():
        has_store = True
if has_store == True:
    print("yes, we have that store")
else:
    print("no, we don't have that store")
```

#### We can use an if statemnt to alter the flow of code. after the word if we include an expression that resolves to true or false. This expression can be an built using comparison and logical operators. If we want one of two things to happen, after out if, we can have an else statement. If there are multiple routes, we can include some elifs in bewteen the if and else

```python
pizza_toppings = ["pepperoni", "mushrooms", "peppers", "tomatoes", "ranch", "cheese"]
topping = input("enter a pizza topping")
for topping in pizza_toppings:
    if topping == "ranch":
        has_ranch = True
if has_ranch == True:
    print("you get an A in the class")
else:
    print("you get a F in the class")
        
```
